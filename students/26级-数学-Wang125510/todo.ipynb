```python
tasks = []


def add_task(tasks, task):
    """把任务添加到待办列表。"""
    tasks.append(task)


def show_tasks(tasks):
    """按添加顺序显示所有任务。"""
    if not tasks:
        print("暂无任务")
        return

    for index, task in enumerate(tasks, start=1):
        print(f"{index}. {task}")
```

```python
tasks = []
add_task(tasks, "学 Python")
add_task(tasks, "写 todo.py")
show_tasks(tasks)
```

```python
allowed = ["add", "show", "exit"]


def check(action):
    for allowed_action in allowed:
        if action == allowed_action:
            return True
    return False


actions = ["add", "show", "exit", "delete", "read_file"]

for action in actions:
    print(action, "允许" if check(action) else "拒绝")
```
