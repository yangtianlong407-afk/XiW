- 定义任务列表、添加函数和查看函数

```python
def add_task(tasks, t):
    tasks.append(t)


def show_tasks(tasks):
    if len(tasks) == 0:
        print("暂无任务")
        return

    for i in range(len(tasks)):
        print(str(i + 1) + ". " + tasks[i])


tasks = []
```

- 直接调用函数，演示连续添加两条任务并按顺序查看

```python
tasks = []
add_task(tasks, "学 Python")
add_task(tasks, "写 todo.py")
show_tasks(tasks)
```
