```python
# Cell 1：定义任务列表和两个函数（只定义，不运行主循环）
tasks = []


def add_task(tasks, task):
    """把 task 追加进列表"""
    tasks.append(task)
    print(f"已添加：{task}")


def show_tasks(tasks):
    """把列表里的任务逐条打印出来"""
    if not tasks:
        print("暂时没有任务。")
        return
    print("当前任务：")
    for i, task in enumerate(tasks, start=1):
        print(f"{i}. {task}")


print("Cell 1 运行完毕：tasks 列表和两个函数已定义好。")
```

```python
# Cell 2：直接调用函数，演示添加和查看
tasks = []
add_task(tasks, "学 Python")
add_task(tasks, "写 todo.py")
show_tasks(tasks)
```
