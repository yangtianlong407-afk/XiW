```python
tasks=[]
def add_task(tasks,task):
    tasks.append(task)
def show_tasks(tasks):
    for i in range(len(tasks)):
        print("{}.{}".format(i+1,tasks[i]))
```

```python
tasks=[]
add_task(tasks,"学Python")
add_task(tasks,"写Python")
show_tasks(tasks)
```
